```python
import qrcode
```

```python
# دریافت اطلاعات شبکه از کاربر
wifi_name = input("نام Wi-Fi را وارد کنید: ")
wifi_password = input("رمز Wi-Fi را وارد کنید: ")
wifi_type = input("نوع امنیت Wi-Fi را وارد کنید (WPA/WPA2/WEP): ")
```

```python
# ساخت رشته استاندارد Wi-Fi برای QR Code
wifi_data = f"WIFI:S:{wifi_name};T:{wifi_type};P:{wifi_password};;"
```

```python
# تبدیل اطلاعات Wi-Fi به QR Code
img = qrcode.make(wifi_data)
```

```python
# ذخیره QR Code به صورت تصویر PNG
img.save("wifi_qrcode.png")
```

```python
print("QR Code با موفقیت ساخته شد!")
```
